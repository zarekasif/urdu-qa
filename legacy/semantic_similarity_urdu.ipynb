```python
from sentence_transformers import SentenceTransformer, util
import torch

embedder = SentenceTransformer('all-MiniLM-L6-v2')
# p = "بیونس نے فریڈرکسبرگ، ٹیکساس میں سینٹ میریز ایلیمنٹری اسکول میں تعلیم حاصل کی، جہاں اس نے رقص کی کلاسوں میں داخلہ لیا۔ اس کی گلوکاری کا ہنر اس وقت دریافت ہوا جب ڈانس انسٹرکٹر ڈارلیٹ جانسن نے ایک گانا گنگنانا شروع کیا اور اس نے اسے ختم کیا، اونچی آواز والے نوٹوں کو مارنے کے قابل۔ بیونس کی موسیقی اور پرفارمنس میں دلچسپی سات سال کی عمر میں ایک اسکول ٹیلنٹ شو جیتنے کے بعد جاری رہی، جس نے جان لینن کا &quot;امیجن&quot; گا کر 15/16 سال کے بچوں کو مات دی۔ 1990 کے موسم خزاں میں، بیونس نے پارکر ایلیمنٹری اسکول میں داخلہ لیا، ہیوسٹن کے ایک میوزک میگنیٹ اسکول، جہاں وہ اسکول کے کوئر کے ساتھ پرفارم کریں گی۔ اس نے ہائی اسکول برائے پرفارمنگ اینڈ ویژول آرٹس اور بعد میں ایلیف ایلسک ہائی اسکول میں بھی تعلیم حاصل کی۔ بیونسے سینٹ جانز یونائیٹڈ میتھوڈسٹ چرچ میں دو سال تک ایک سولوسٹ کے طور پر کوئر کی رکن بھی تھیں۔"
# corpus = p.split('۔')
# corpus_embeddings = embedder.encode(corpus, convert_to_tensor=True)


import pandas as pd
# import os
# import csv
file = pd.read_csv("updated dataset.csv")
```

```python
correct = 0
for i in range(10):
    p = file.Paragraphs[i]
    q = file.Questions[i]
    a = file.Answers[i]
    corpus = p.split('.')
    corpus_embeddings = embedder.encode(corpus, convert_to_tensor=True)
    top_k = min(1, len(corpus))
    q_embedding = embedder.encode(q, convert_to_tensor=True)
        # We use cosine-similarity and torch.topk to find the highest 5 scores
    cos_scores = util.cos_sim(q_embedding, corpus_embeddings)[0]
    top_results = torch.topk(cos_scores, k=top_k)
#     print("\n\n======================\n\n")
#     print("Query:", q)
#     print("\nTop 1 most similar sentences in corpus:")

    for score, idx in zip(top_results[0], top_results[1]):
#         print(corpus[idx], "(Score: {:.4f})".format(score))
        if a in corpus[idx]:
            #print(q,a)
            correct += 1
            
        else:
            pass


print(correct)
# print((correct/23000)*100)
```
