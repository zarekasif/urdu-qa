```python
# Make predictions with the model
from rouge import Rouge
ROUGE = Rouge()
correct = 0
lines = 0
loop = 100
Rouge = []
F1 = []
EM = []
for i in range(loop):

    p = file.Paragraphs[i]
    q = file.Questions[i]
    a = file.Answers[i]
    to_predict = [
        {
            "context": p,
            "qas": [
                {
                    "question": q ,
                    "id": "0",
                }
            ],
        }
    ]
    answers, probabilities = model.predict(to_predict)
    res = answers[0]["answer"]
#     print(res)
    if a in res:
#       print("correct ans is :",a)
      correct += 1

    print("Running line is: ", lines)
    print("Actual answer is",a)
#     print("Running line is: ", lines)
    lines += 1
    answer = max(res)
    print("Pridicted answer is ",max(res))
       
#     print("Printing F1, Exact Match, Rouge")
    
    
    temp = ROUGE.get_scores(a, answer)[0]["rouge-1"]["f"]
    Rouge.append(temp)
    print("Printing Rouge score",temp)
#     print()
#     Computing exact match 
    if a == answer:
        EM.append(answer)

#     Computing F1 score
#     a = "سے زیادہ فروخت ہونے لڑکیوں گروپوں میں"
#     fin_max = "دنیا اب سب سے زیادہ فروخت ہونے لڑکیوں گروپوں میں"
#     F1 = []
    an = a.split(" ")
    fn = answer.split(" ")
#     print(an)
#     print(fn)
    TP = 0
    FP = 0
    TN = 0
    Precision = 0
    Recall = 0
    if a in answer:
        TP = len(an)
        FP = len(fn)-len(an)
        TN = len(an)
        Precision = TP/(TP+FP)
        Recall = TP/TP
#         print(TP,FP,TN)
#         print(Precision, Recall)

        f1 = 2*(Precision*Recall)/(Precision+Recall)
        F1.append(f1)
        print("F1 is ",f1)
    else:
        F1.append(0)
    print("//////////////////////////////////////////////")
#     print(p)
#     print(q)
#     print(a)

    print("Rouge score till now: ", sum(Rouge)/len(Rouge))
    print("Rouge List is ",Rouge)
    print("total Exact match till now",len(EM))
    print("EM List is ",EM)
    print("Average of F1 till now ",sum(F1)/len(F1))
    print(" ")
    print("Line END")
    print(" ")
    print(" ")
    print("F1 List is ",F1)



    print("Accuracy till now: ", (correct/lines)*100)


```
